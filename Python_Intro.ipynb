## B · Funktionen, Fehler und Objektorientierung
Lernziel: wiederverwendbare Berechnungen schreiben, Randfälle prüfen und Zustand in Objekten bündeln. **Bezug: Folien 14–19.**


````python
def netto(preis: float, rabatt: float = 0.0) -> float:
    """Berechne einen rabattierten Preis. Rabatt liegt zwischen 0 und 1."""
    if preis < 0 or not 0 <= rabatt <= 1:
        raise ValueError("Preis oder Rabatt unzulässig")
    return preis * (1 - rabatt)

assert netto(100, 0.1) == 90
assert netto(100, 1) == 0
print(netto(80, rabatt=0.25))
````

`return` liefert einen Wert zurück, `print` zeigt ihn nur an. Funktionsparameter sind lokale Namen. Typannotationen dokumentieren Erwartungen, erzwingen sie aber nicht automatisch. `assert` hilft bei kleinen Tests. Die fachliche Prüfung erfolgt explizit mit `if` und `raise`.

**Debugging:** Fehlermeldung unten im Traceback lesen, dann die eigene Codezeile prüfen. Reproduziere das Problem mit möglichst wenigen Eingaben. Unbekannte Namen verursachen oft `NameError`, ungeeignete Typen `TypeError`, ungeeignete Werte `ValueError`.


````python
for eingabe in ["12.5", "zwölf"]:
    try:
        print(float(eingabe))
    except ValueError:
        print(f"{eingabe!r} ist keine Zahl im erwarteten Format")

try:
    netto(100, 1.5)
except ValueError as fehler:
    print("Erwarteter Fehler:", fehler)
````

````python
class Produkt:
    def __init__(self, name, preis):
        if preis < 0:
            raise ValueError("Preis muss nichtnegativ sein")
        self.name = name
        self.preis = preis

    def umsatz(self, menge):
        if menge < 0:
            raise ValueError("Menge muss nichtnegativ sein")
        return self.preis * menge

sensor = Produkt("Sensor", 12.5)
kabel = Produkt("Kabel", 3.0)
print(sensor.name, sensor.umsatz(4))
print(kabel.name, kabel.umsatz(4))
````

Die Klasse ist der Bauplan, `sensor` und `kabel` sind Instanzen. `__init__` initialisiert den Zustand, `self` bezeichnet die jeweilige Instanz. `name` und `preis` sind Attribute, `umsatz` ist eine Methode. Die Attribute sind hier öffentlich. In produktivem Code müssten spätere unzulässige Änderungen ebenfalls berücksichtigt werden.

**Komposition** kombiniert Objekte, z. B. eine Liste von Produkten als Warenkorb. **Vererbung** spezialisiert eine Klasse. Beides ist optional für diesen Einstieg. Viele Data-Science-Bibliotheken stellen bereits fertige Klassen bereit.

## D · Suchen, Sortieren und Einfügen
Lernziel: Algorithmen nach Voraussetzungen und Aufwand unterscheiden. **Bezug: Folien 24–30.**

| Struktur / Operation | Aufwand | Bedingung / Einordnung |
|---|---|---|
| list: Indexzugriff | O(1) | Index bekannt |
| list: lineare Suche | O(n) | keine Sortierung nötig |
| list: append | amortisiert O(1) | gelegentliche Vergrößerung |
| list: insert / Entfernen in der Mitte | O(n) | Elemente verschieben |
| sortierte list: binäre Suche | O(log n) | schneller Indexzugriff |
| sortierte list: Einfügen | O(n) | Suche plus Verschieben |
| dict / set: Zugriff und Einfügen | durchschnittlich O(1), Worst Case O(n) | Hashing und Kollisionen |
| Stack per list: append / pop am Ende | amortisiert O(1) | leerer Stack braucht Sonderfall |
| deque: append / popleft | O(1) | Operationen an den Enden |
| balancierter Suchbaum: Suche / Einfügen | O(log n) | Höhe logarithmisch |
| unbalancierter Suchbaum: Suche / Einfügen | Worst Case O(n) | kann zur Kette werden |
| sorted / list.sort | Worst Case O(n log n) | stabile Sortierung |

Big O beschreibt das Wachstum, nicht die gemessene Zeit. `n` ist die Anzahl der Elemente. Kleine Beispiele dienen dem Verständnis, nicht als belastbarer Benchmark.


````python
def linear_suche(werte, gesucht):
    for i, wert in enumerate(werte):
        if wert == gesucht:
            return i
    return -1

def binaer_suche(werte, gesucht):
    links, rechts = 0, len(werte) - 1
    while links <= rechts:
        mitte = (links + rechts) // 2
        if werte[mitte] == gesucht:
            return mitte
        if werte[mitte] < gesucht:
            links = mitte + 1
        else:
            rechts = mitte - 1
    return -1

assert linear_suche([7, 2, 9], 2) == 1
assert binaer_suche([2, 5, 8, 11, 17], 11) == 3
assert binaer_suche([], 11) == -1
assert binaer_suche([2, 5], 3) == -1
print("Suchtests bestanden")
````

Bei Duplikaten liefert diese binäre Suche irgendeinen passenden Index. Für den ersten möglichen Index eignet sich `bisect_left`. Ein Rückgabewert `-1` bedeutet hier „nicht gefunden“ und darf nicht ungeprüft als Listenindex dienen.


````python
from bisect import bisect_left, insort
werte_sortiert = [2, 5, 8, 11, 17]
i = bisect_left(werte_sortiert, 6)
print("Einfügeposition für 6:", i)
insort(werte_sortiert, 6)
print(werte_sortiert)
werte_unsortiert = [8, 2, 5]
print(sorted(werte_unsortiert), werte_unsortiert)
werte_unsortiert.sort()
print(werte_unsortiert)
````

````python
def insertion_sort(werte):
    a = list(werte)
    for i in range(1, len(a)):
        x, j = a[i], i - 1
        while j >= 0 and a[j] > x:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = x
    return a

for beispiel in [[], [1], [5, 2, 4], [3, 3, -1], [1, 2, 3]]:
    assert insertion_sort(beispiel) == sorted(beispiel)
print(insertion_sort([5, 2, 4]))
````

### Optional: ein einfacher binärer Suchbaum
Jeder Knoten speichert einen Wert, einen linken und einen rechten Teilbaum. Kleinere Werte liegen links, größere rechts. In diesem Beispiel ignorieren wir Duplikate. Der Baum balanciert sich nicht selbst. Der Aufwand hängt von der Höhe ab. Wir verwenden iterative Operationen, damit der Aufrufstack nicht das erste Hindernis ist.


````python
class Knoten:
    def __init__(self, wert):
        self.wert = wert
        self.links = None
        self.rechts = None

    def einfuegen(self, wert):
        aktuell = self
        while True:
            if wert == aktuell.wert:
                return
            seite = "links" if wert < aktuell.wert else "rechts"
            kind = getattr(aktuell, seite)
            if kind is None:
                setattr(aktuell, seite, Knoten(wert))
                return
            aktuell = kind

    def enthaelt(self, gesucht):
        aktuell = self
        while aktuell is not None:
            if gesucht == aktuell.wert:
                return True
            aktuell = aktuell.links if gesucht < aktuell.wert else aktuell.rechts
        return False

baum = Knoten(8)
for wert in [3, 10, 1, 6]:
    baum.einfuegen(wert)
assert baum.enthaelt(6)
assert not baum.enthaelt(7)
print("6 enthalten:", baum.enthaelt(6))
````

## E · Data Science: Daten verstehen und visualisieren
Lernziel: mit NumPy rechnen, CSV-Daten mit pandas untersuchen und eine fachliche Frage visualisieren. **Bezug: Folien 31–36.**

**Fall:** Ein fiktiver Händler beobachtet unabhängige Verkaufssituationen. Wir wollen Zusammenhänge beschreiben und später Absatz vorhersagen. Das ist kein Zeitreihen- oder Kundendatensatz.

| Spalte | Bedeutung |
|---|---|
| id | eindeutige Verkaufssituation |
| werbung_eur | vorab geplantes Werbebudget in Euro |
| rabatt | vorab bekannter Rabatt als Anteil, z. B. 0.1 |
| kanal | vorab bekannter Vertriebskanal |
| absatz | beobachtete Stückzahl, später unser Zielwert |
| umsatz_eur | nachträglich berechnet aus Absatz und Stückpreis |

Wir erzeugen 240 Beobachtungen, ergänzen absichtlich 12 fehlende Budgetwerte und 4 identische Duplikate. Der Zufallsstartwert macht den Datensatz wiederholbar. **Der Generator ist ein Lernmodell und keine Aussage über reale Absatztreiber.**


````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from io import StringIO
from pathlib import Path
from sklearn.model_selection import train_test_split

rng = np.random.default_rng(42)
n = 240
werbung = rng.uniform(50, 1000, n)
rabatte = rng.choice([0.0, 0.05, 0.10, 0.20], n)
kanaele = rng.choice(["Web", "Shop", "Partner"], n)
kanal_effekt = pd.Series(kanaele).map({"Web": 12, "Shop": 0, "Partner": 6}).to_numpy()
absatz = np.maximum(0, np.rint(20 + 0.07 * werbung + 90 * rabatte
                              + kanal_effekt + rng.normal(0, 9, n))).astype(int)
daten = pd.DataFrame({"id": np.arange(1, n + 1), "werbung_eur": werbung.round(2),
                      "rabatt": rabatte, "kanal": kanaele, "absatz": absatz})
daten["umsatz_eur"] = (daten["absatz"] * 20 * (1 - daten["rabatt"])).round(2)
fehlende = rng.choice(n, 12, replace=False)
daten.loc[fehlende, "werbung_eur"] = np.nan
roh = pd.concat([daten, daten.iloc[:4]], ignore_index=True)
# In sich geschlossenes CSV-Beispiel, kein Download nötig:
csv_text = roh.to_csv(index=False)
df = pd.read_csv(StringIO(csv_text))
print("Zeilen und Spalten:", df.shape)
display(df.head())
````

NumPy hat die Zahlenfelder erzeugt und die Absatzformel elementweise ausgewertet. pandas macht daraus eine Tabelle. In einem echten Projekt ersetzt man `StringIO(csv_text)` zum Beispiel durch `"verkauf.csv"`. Ein Semikolon als Trennzeichen braucht `sep=";"`, ein deutsches Dezimalformat gegebenenfalls `decimal=","`.


````python
df.info()
print("Fehlende Werte:")
print(df.isna().sum())
print("Doppelte IDs:", df["id"].duplicated().sum())
assert len(df) == 244
# Wir wissen aus dem Generator, dass diese Wiederholungen identisch sind.
sauber = df.drop_duplicates(subset="id").copy()
assert len(sauber) == 240
assert sauber["id"].is_unique
assert sauber["rabatt"].between(0, 1).all()
assert (sauber["absatz"] >= 0).all()
````

Die ID hilft, bekannte identische Duplikate zu entfernen. Bei echten Daten müsste man vorab klären, ob wiederholte IDs Korrekturen oder unterschiedliche Ereignisse darstellen. Wir entfernen die Duplikate **vor** dem Split, damit identische Fälle nicht in Training und Test landen.

Nun trennen wir die Testdaten ab. Alle explorativen Grafiken und Entscheidungen zur Modellierung beziehen sich ab hier auf den Trainingsanteil. Fehlende Merkmale werden erst innerhalb der Modellpipeline behandelt.


````python
train_df, test_df = train_test_split(sauber, test_size=0.25, random_state=42)
train_df = train_df.copy()
test_df = test_df.copy()
assert set(train_df["id"]).isdisjoint(set(test_df["id"]))
print("Train:", len(train_df), "Test:", len(test_df))
display(train_df.describe(include="all"))
display(train_df.groupby("kanal").agg(faelle=("id", "size"),
                                      absatz_mittel=("absatz", "mean"),
                                      umsatz_summe=("umsatz_eur", "sum")))
````

````python
hoher_absatz = train_df.loc[train_df["absatz"] >= 100,
                            ["id", "kanal", "absatz"]]
print("Trainingsfälle mit mindestens 100 Stück:", len(hoher_absatz))
preise_np = np.array([10.0, 20.0, 15.0])
mengen_np = np.array([2, 1, 3])
print("Umsätze:", preise_np * mengen_np)
print("Mittlerer Umsatz:", (preise_np * mengen_np).mean())
````

````python
fig, ax = plt.subplots(figsize=(7, 4))
for kanal, teil in train_df.groupby("kanal"):
    ax.scatter(teil["werbung_eur"], teil["absatz"], label=kanal, alpha=0.7)
ax.set(xlabel="Werbebudget (Euro)", ylabel="Absatz (Stück)",
       title="Synthetische Trainingsdaten: Budget und Absatz")
ax.legend(title="Kanal")
fig.tight_layout()
plt.show()
````

## F · Ein erstes Vorhersagemodell
Lernziel: Features und Zielwert unterscheiden, Vorverarbeitung kapseln und die Güte im Vergleich zu einer Baseline beurteilen. **Bezug: Folien 37–39.**

Wir sagen **Absatz in Stück** voraus. Als Eingangsmerkmale nutzen wir nur vorab bekannte Größen: Werbebudget, Rabatt und Kanal. `id` hat keine fachliche Vorhersagefunktion. `umsatz_eur` enthält bereits den Absatz und wäre eine undichte Information über das Ziel.

Die Pipeline füllt numerische Lücken mit einem im Training berechneten Median und kodiert den Kanal. Eine lineare Regression lernt eine einfache additive Beziehung. `DummyRegressor` sagt als Baseline immer den Median des Trainingszielwerts voraus. Das ist für den MAE eine passende konstante Baseline.

**MAE:** Mittelwert der absoluten Abweichungen. MAE = 8 bedeutet beispielsweise durchschnittlich 8 Stück Abweichung pro Fall, nicht 8 Prozent und keine garantierte Fehlergrenze.


````python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LinearRegression
from sklearn.dummy import DummyRegressor
from sklearn.model_selection import KFold, cross_val_score
from sklearn.metrics import mean_absolute_error

merkmale = ["werbung_eur", "rabatt", "kanal"]
X_train, y_train = train_df[merkmale], train_df["absatz"]
X_test, y_test = test_df[merkmale], test_df["absatz"]
vorverarbeitung = ColumnTransformer([
    ("zahlen", SimpleImputer(strategy="median"), ["werbung_eur", "rabatt"]),
    ("kategorien", OneHotEncoder(handle_unknown="ignore", drop="first"), ["kanal"])
])
modell = Pipeline([("vorverarbeitung", vorverarbeitung),
                   ("regression", LinearRegression())])
baseline = DummyRegressor(strategy="median")
cv = KFold(n_splits=5, shuffle=True, random_state=42)
cv_mae = -cross_val_score(modell, X_train, y_train,
                          cv=cv, scoring="neg_mean_absolute_error")
cv_basis = -cross_val_score(baseline, X_train, y_train,
                            cv=cv, scoring="neg_mean_absolute_error")
print("CV-MAE Modell:", cv_mae.round(2), "Mittel:", round(cv_mae.mean(), 2))
print("CV-MAE Baseline:", round(cv_basis.mean(), 2))
````

Cross-Validation teilt nur die Trainingsdaten wiederholt in Lern- und Validierungsanteile. Die Pipeline lernt die Imputation in jedem Fold neu. scikit-learn maximiert Scores und verwendet deshalb hier einen negativen Fehlerwert, dessen Vorzeichen wir umdrehen. Vor dem nächsten Schritt alle Modellentscheidungen abschließen. Danach folgt die einmalige Testbewertung.


````python
modell.fit(X_train, y_train)
baseline.fit(X_train, y_train)
vorhersage = modell.predict(X_test)
basis_vorhersage = baseline.predict(X_test)
mae_modell = mean_absolute_error(y_test, vorhersage)
mae_basis = mean_absolute_error(y_test, basis_vorhersage)
ergebnis = pd.DataFrame({"Modell": ["Median-Baseline", "Lineare Regression"],
                         "Test-MAE (Stück)": [mae_basis, mae_modell]})
display(ergebnis.round(2))
assert np.isfinite(vorhersage).all()
assert mae_modell < mae_basis, "Beim festen Lerndatensatz erwarten wir eine Verbesserung."
print("Die Modellgüte gilt für diese synthetische Datenverteilung.")
````

````python
fig, ax = plt.subplots(figsize=(5, 5))
ax.scatter(y_test, vorhersage, alpha=0.7)
grenzen = [min(y_test.min(), vorhersage.min()), max(y_test.max(), vorhersage.max())]
ax.plot(grenzen, grenzen, "--", color="gray", label="Perfekte Vorhersage")
ax.set(xlabel="Tatsächlicher Absatz (Stück)", ylabel="Vorhergesagter Absatz (Stück)",
       title="Abschließende Testbewertung")
ax.legend()
fig.tight_layout()
plt.show()

neue_faelle = pd.DataFrame({"werbung_eur": [300, 700], "rabatt": [0.1, 0.05],
                           "kanal": ["Web", "Shop"]})
print("Neue Vorhersagen:", modell.predict(neue_faelle).round(1))
````

Das Modell kann nichtganzzahlige oder bei ungeeigneten Eingaben sogar negative Stückzahlen schätzen. Für echte Anwendungen müsste man Modellannahmen, zulässige Eingaben und fachliche Randbedingungen prüfen. Ein Vergleich auf demselben Testset nach vielen Anpassungen wäre optimistisch. Für weitere Experimente nur die Trainings-Cross-Validation verwenden und einen neuen unberührten Test für eine spätere Schlussbewertung vorsehen.

**Grenzen:** künstliche Daten, einfache additive Generierungsregel, zufällig fehlende Werte, keine zeitliche Entwicklung und keine realen Kundengruppen. Der feste Zufallsstartwert allein garantiert noch keine identischen Resultate über jede Bibliotheksversion oder Hardware hinweg.

### Lösung B1 und B2


````python
def durchschnitt(werte):
    if len(werte) == 0:
        raise ValueError("Mindestens ein Wert erforderlich")
    return sum(werte) / len(werte)

assert durchschnitt([2, 4, 6]) == 4
assert durchschnitt([7]) == 7
assert durchschnitt([-4, 2]) == -1
try:
    durchschnitt([])
except ValueError:
    pass
else:
    raise AssertionError("Die leere Liste muss einen Fehler auslösen")

class Warenkorb:
    def __init__(self):
        self.produkte = []
    def hinzufuegen(self, produkt):
        self.produkte.append(produkt)
    def gesamt(self):
        return sum(p.preis for p in self.produkte)

korb1, korb2 = Warenkorb(), Warenkorb()
korb1.hinzufuegen(Produkt("Sensor", 12.5))
korb1.hinzufuegen(Produkt("Kabel", 3))
assert korb1.gesamt() == 15.5
assert korb2.gesamt() == 0
print("Funktion und Warenkörbe geprüft")
````
